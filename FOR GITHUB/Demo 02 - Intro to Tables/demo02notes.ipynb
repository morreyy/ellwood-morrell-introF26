```python
# Creates a new table with only the selected column
flavor_table = cones.select('Flavor')
```

```python
# Notice the old table is unmodified
cones.show()
flavor_table.show()
```

```python
# Since we did not store this anywhere, we can see the output but cannot use it for anything
cones.select('Flavor', 'Price')
```

```python
# The old table is unmodified (and the 'Flavor'/'Price' one is lost)
cones.show()
```

```python
# Creates a new table with the specified column missing
cones.drop('Price')
```

```python
cones.show()
```

```python
cones_without_price = cones.drop('Price')
cones_without_price
```

```python
# Selects all rows where 'Flavor' is equal to 'chocolate'
cones.where('Flavor', 'chocolate')
```

```python
# Sorts the table by price
cones.sort('Price')
```

```python
cones.sort('Price', descending=True)
```

```python
cones.sort('Flavor', descending=True)
```

```python
# You can reuse variable names when you need to make multiple edits.
chocolate_table = cones.where('Flavor', 'chocolate')
chocolate_table = chocolate_table.drop('Flavor')
chocolate_table.show()
```

### A more interesting table

```python
# From https://github.com/erikgregorywebb/datasets/blob/master/nba-salaries.csv
nba = Table.read_table('nba_salaries.csv')
nba
```

```python
point_guards = nba.where('position', 'PG')
point_guards_2020 = point_guards.where('season', 2020)
```

```python
point_guards_2020.show(10)
```

```python
point_guards.num_rows
```

```python
point_guards_2020.num_rows
```

```python
point_guards_2020 = point_guards_2020.drop('position')
```

```python
# We forgot to assign the table in the previous cell to a variable
point_guards_2020
```

```python
# Now our changed table is stored in point_guards
point_guards_2020 = point_guards_2020.drop('rank', 'position', 'season')
point_guards_2020.show(10)
```

```python
# Here's another way to get the columns you want by directly selecting them
point_guards_alt = nba.where('position', 'PG').where('season', 2020)
point_guards_alt = point_guards_alt.select('name', 'team', 'salary')
point_guards_alt.show(10)
```

```python
point_guards_2020.sort('salary').show(10)
```

```python
point_guards_2020.sort('salary', descending=True).show(10)
```
